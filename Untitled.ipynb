## Regression

```python
from scipy.optimize import minimize

def bellman(q, shape, obs, gamma=0.95):
    q = q.reshape(shape)
    r = obs['rewards']
    s0 = obs['state0']
    s1 = obs['state1']
    a = obs['action']
    s01, s02 = np.array(s0).T
    s11, s12 = np.array(s1).T
    qs0 = q[s01, s02, a]
    vs1 = np.max(q[s11, s12], axis=-1)
    return np.sum((qs0 - gamma * vs1 - r)**2)
    
```

```python
shape = (env.n_cell + (env.action_space.n,))
q0 = np.random.normal(size=shape)
q_opt = minimize(bellman, q0.flatten(), args=(shape, env.uniform_obs._buffers), method="BFGS")
q_opt
fig, ax = plt.subplots()
im = ax.imshow(np.max(q_opt.x.reshape(shape), axis=-1))
fig.colorbar(im, ax=ax, location='right', anchor=(0, 0.3), shrink=1)
```
